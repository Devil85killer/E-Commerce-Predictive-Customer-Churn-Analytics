```python
import pandas as pd
import numpy as np
import urllib

from sqlalchemy import create_engine

print("Libraries imported successfully!")
```

```python
server = r"DEVIL\SQLEXPRESS"
database = "ecommerce"

params = urllib.parse.quote_plus(
    f"DRIVER={{ODBC Driver 17 for SQL Server}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    f"Trusted_Connection=yes;"
)

engine = create_engine(
    "mssql+pyodbc:///?odbc_connect=" + params
)

print("SQL Server connection successful!")
```

```python
rfm_query = """
SELECT
    customer_unique_id,
    recency,
    frequency,
    monetary
FROM dbo.Customer_RFM
"""

rfm = pd.read_sql(
    rfm_query,
    engine
)

print("RFM Shape:", rfm.shape)

rfm.head()
```

```python
transactions_query = """
SELECT
    customer_unique_id,
    order_id,
    order_purchase_timestamp,
    total_order_value
FROM dbo.Customer_Transactions
"""

transactions = pd.read_sql(
    transactions_query,
    engine
)

transactions["order_purchase_timestamp"] = pd.to_datetime(
    transactions["order_purchase_timestamp"]
)

print("Transactions Shape:", transactions.shape)

transactions.head()
```

```python
print(
    "First Order:",
    transactions["order_purchase_timestamp"].min()
)

print(
    "Last Order:",
    transactions["order_purchase_timestamp"].max()
)
```

```python
print("RFM rows:", len(rfm))
print("Transaction rows:", len(transactions))

print("\nRFM missing values:")
print(rfm.isnull().sum())

print("\nTransaction missing values:")
print(transactions.isnull().sum())
```
